# Regression

1. What is Simple Linear Regression?

Ans - Simple Linear Regression is a statistical method used to model and analyze the relationship between two continuous variables:

a. Independent Variable ($X$): The predictor or explanatory variable.

b. Dependent Variable ($Y$): The outcome or response variable.It assumes a linear relationship and models it using a straight-line equation:$$Y = mX + c + \epsilon$$

2. What are the key assumptions of Simple Linear Regression?

Ans - For the model's estimates to be valid and reliable, four core assumptions must hold:

a. Linearity: The relationship between $X$ and $Y$ must be linear.

b. Independence: The observations/residuals must be independent of each other.

c. Homoscedasticity: The variance of the residual errors must be constant across all levels of $X$.

d. Normality: The residual errors must be normally distributed (especially important for hypothesis testing).

3. What does the coefficient $m$ represent in the equation $Y=mX+c$?

Ans - The coefficient $m$ represents the slope of the regression line. It quantifies the change in the expected value of the dependent variable ($Y$) for every one-unit increase in the independent variable ($X$).If $m > 0$, there is a positive relationship.If $m < 0$, there is a negative relationship.

4. What does the intercept $c$ represent in the equation $Y=mX+c$?

Ans - The intercept $c$ (often written as $\beta_0$) represents the y-intercept. It is the predicted value of the dependent variable ($Y$) when the independent variable ($X$) is exactly equal to zero.

5. How do we calculate the slope $m$ in Simple Linear Regression?

Ans - The slope $m$ is calculated using the formulas derived from minimizing the sum of squared errors:

$$m = \frac{\sum_{i=1}^{n} (X_i - \bar{X})(Y_i - \bar{Y})}{\sum_{i=1}^{n} (X_i - \bar{X})^2} = r \frac{s_y}{s_x}
$$Where:$\bar{X}$ and $\bar{Y}$ are the sample means.$r$ is the Pearson correlation coefficient between $X$ and $Y$.$s_x$ and $s_y$ are the sample standard deviations.

6. What is the purpose of the least squares method in Simple Linear Regression?

Ans - The purpose of the Ordinary Least Squares (OLS) method is to find the unique line of "best fit." It achieves this by minimizing the sum of the squared differences (residuals) between the actual observed values ($Y_i$) and the predicted values ($\hat{Y}_i$) generated by the line:

$$\text{Minimize } \sum_{i=1}^{n} (Y_i - \hat{Y}_i)^2$$

7. How is the coefficient of determination ($R^2$) interpreted in Simple Linear Regression?

Ans - The coefficient of determination, $R^2$, measures the proportion of the total variance in the dependent variable ($Y$) that is explained or predicted by the independent variable ($X$).

An $R^2$ of 0.85 means that 85% of the variation in $Y$ is explained by $X$, while the remaining 15% is due to unexplained variation or random error. It ranges from 0 to 1.

8. What is Multiple Linear Regression?

Ans - Multiple Linear Regression extends simple linear regression by modeling the relationship between one continuous dependent variable ($Y$) and two or more independent variables ($X_1, X_2, \dots, X_k$). Its general equation is:

$$Y = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + \dots + \beta_k X_k + \epsilon$$

9. What is the main difference between Simple and Multiple Linear Regression?

Ans - Simple Linear Regression uses exactly one independent variable to predict $Y$.Multiple Linear Regression uses two or more independent variables simultaneously to predict $Y$, allowing the model to account for multiple confounding factors.

10. What are the key assumptions of Multiple Linear Regression?

Ans - It shares the same foundational assumptions as simple linear regression, with one critical addition:

a. Linearity: The relationship between the features and $Y$ is linear.Independence of residuals.Homoscedasticity of residuals.Normality of residuals.

b. No Multicollinearity (Addition): The independent variables must not be highly correlated with each other.

11. What is heteroscedasticity, and how does it affect the results of a Multiple Linear Regression model?

Ans - Heteroscedasticity occurs when the variance of the residual errors is not constant but changes across different levels of the predictor variables.Effects:It does not bias the regression coefficients (they remain unbiased).However, it invalidates standard errors. Because standard errors are miscalculated, your p-values, confidence intervals, and hypothesis tests ($t$-tests and $F$-tests) become completely unreliable, leading to false conclusions about feature significance.

12. How can you improve a Multiple Linear Regression model with high multicollinearity?

Ans - Multicollinearity makes it difficult to isolate the individual effect of a single feature. You can mitigate it by:

a. Removing Variables: Drop one of the highly correlated variables using a Variance Inflation Factor (VIF) threshold (typically drop if $\text{VIF} > 5 \text{ or } 10$).

b. Feature Combination: Combine the correlated variables into a single index or score.

c. Dimensionality Reduction: Use methods like Principal Component Analysis (PCA) to create orthogonal (uncorrelated) components.

d. Regularization: Switch to Ridge or Lasso regression, which penalizes large coefficients.

13. What are some common techniques for transforming categorical variables for use in regression models?

Ans - Regression algorithms require numeric inputs. Categorical variables can be transformed using:

a. One-Hot Encoding (Dummy Encoding): Converts a categorical column into multiple binary columns ($0$ or $1$).

Note: To avoid the dummy variable trap (multicollinearity), drop one category ($k-1$ columns for $k$ categories).

b. Label/Ordinal Encoding: Assigns sequential integers ($1, 2, 3...$) to categories. This should only be used if the categorical variable has a natural, inherent order (e.g., Low, Medium, High).

14. What is the role of interaction terms in Multiple Linear Regression?

Ans - An interaction term is created by multiplying two independent variables together ($X_1 \times X_2$). Its role is to capture situations where the effect of one predictor variable on $Y$ depends on the level/value of another predictor variable, rather than assuming their effects are simply additive.

15. How can the interpretation of intercept differ between Simple and Multiple Linear Regression?

Ans - In Simple Linear Regression, the intercept is the baseline value of $Y$ when the single predictor $X=0$.In Multiple Linear Regression, the intercept is the expected value of $Y$ when all included independent variables are simultaneously equal to zero ($X_1=0, X_2=0, \dots$). If setting all variables to zero is physically or contextually impossible (e.g., predicting house price where square footage, bedrooms, and age are all $0$), the intercept serves purely as a mathematical anchor and lacks real-world meaning.

16. What is the significance of the slope in regression analysis, and how does it affect predictions?

Ans - The slope indicates both the direction and the magnitude of the relationship between variables.Significance: It tells us how sensitive our target variable is to shifts in the features.

Effect on Predictions: The slope acts as a direct multiplier. If the slope for a feature is $\beta_1 = 5.2$, any time a new data point increases by $1$ unit in that feature, the model automatically adds $5.2$ to its final prediction.

17. How does the intercept in a regression model provide context for the relationship between variables?

Ans - The intercept establishes the baseline starting point of the system. It anchors the model's line or hyperplane in space. Without an intercept, the regression line would be forced to pass through the origin $(0,0)$, which falsely assumes that if the inputs are zero, the output must also be zero (which can severely bias predictions).

18. What are the limitations of using $R^2$ as a sole measure of model performance?

Ans - The Inflation Flaw: $R^2$ will never decrease when you add more independent variables, even if those variables are completely irrelevant noise. This rewards overfitted, overly complex models.

Context Blindness: A high $R^2$ doesn't mean a model is structurally correct; you could still have violated crucial assumptions (like homoscedasticity) or missed non-linear relationships.

19. How would you interpret a large standard error for a regression coefficient?

Ans - A large standard error means that the estimate of that specific coefficient has high variability and low precision. It implies that if you were to sample different subsets of data, the calculated coefficient would fluctuate wildly. Consequently, a large standard error typically results in a low $t$-statistic and a high p-value, meaning the feature is likely statistically insignificant in predicting $Y$.

20. How can heteroscedasticity be identified in residual plots, and why is it important to address it?

Ans -

Identification: In a plot of Residuals vs. Fitted Values, heteroscedasticity manifests visually as a distinct "fan", "cone", or "funnel" shape, where the spread of the residuals expands or contracts as predictions get larger.

Why address it: Because failing to address it invalidates your p-values and confidence calculations, potentially tricking you into executing business or engineering decisions based on features that aren't actually reliably significant.

21. What does it mean if a Multiple Linear Regression model has a high $R^2$ but low adjusted $R^2$?

Ans - This is a textbook sign of overfitting via feature stuffing. It indicates that the model contains too many irrelevant, non-predictive independent variables. While $R^2$ goes up artificially with every added column, Adjusted $R^2$ penalizes the model for adding useless features. A large gap between the two tells you to strip away the noise.

22. Why is it important to scale variables in Multiple Linear Regression?

Ans - While scaling is not mathematically mandatory for finding the absolute baseline OLS solution, it is vital for several reasons:

a. Interpretability: If features are on different scales (e.g., age in years vs. annual income in millions), their raw coefficients cannot be directly compared to evaluate feature importance. Scaling forces coefficients to be on a leveled playing field.

b. Optimization: If you utilize Gradient Descent to train your regression model, scaling prevents the optimization path from oscillating inefficiently, accelerating convergence.

c. Regularization: If you apply Ridge or Lasso penalties, scaling is strictly required; otherwise, variables with naturally larger scales will be penalized unfairly.

23. What is polynomial regression?

Ans - Polynomial Regression is a form of regression analysis in which the relationship between the independent variable $X$ and the dependent variable $Y$ is modeled as an $n$-th degree polynomial.While it models a non-linear relationship in the data space, it is still considered a form of linear regression because the parameters/coefficients ($\beta_i$) enter the equation linearly.

24. How does polynomial regression differ from linear regression?

Ans - Linear Regression fits a straight line, plane, or hyperplane to the data, assuming a constant rate of change.Polynomial Regression introduces curves by using powers of the independent variables ($X^2, X^3, \dots$). This allows it to capture fluctuating upward and downward trends.

25. When is polynomial regression used?

Ans - It is used when the true relationship between variables is fundamentally non-linear, curvilinear, or exhibits structural shifts—such as modeling the progression of an epidemic, the growth rate of tissues, or market yield curves.

26. What is the general equation for polynomial regression?

Ans - The general equation for a single variable transformed to an $n$-th degree polynomial is:$$Y = \beta_0 + \beta_1 X + \beta_2 X^2 + \beta_3 X^3 + \dots + \beta_n X^n + \epsilon$$

27. Can polynomial regression be applied to multiple variables?

Ans - Yes. When applied to multiple independent variables ($X_1, X_2$), the model includes not only their individual polynomial powers ($X_1^2, X_2^2$) but also cross-product interaction terms (e.g., $X_1 X_2$), mapping a complex multi-dimensional curved surface.

28. What are the limitations of polynomial regression?

Ans -

a. Severe Overfitting: If the degree ($n$) is set too high, the model will aggressively curve to hit every single noisy training point, destroying its ability to generalize to new data.

b. Extreme Boundary Instability (Runge's Phenomenon): Polynomials behave wildly outside the range of your training data. Extrapolating predictions beyond your minimum or maximum observed values is highly unreliable.

29. What methods can be used to evaluate model fit when selecting the degree of a polynomial?

Ans -

a. Cross-Validation (k-fold): Split your data, train on a subset, and evaluate performance on a validation holdout. Watch for the degree where validation error begins to rise (the sweet spot before overfitting).

b. Information Criteria (AIC / BIC): Metrics that evaluate model fit while strictly penalizing higher parameter counts. Lower AIC/BIC values point to the optimal degree.Adjusted $R^2$ Check: Look for where the Adjusted $R^2$ plateaus or starts declining.

30. Why is visualization important in polynomial regression?

Ans - Because polynomial models can easily bend into extreme shapes, simply looking at metrics like $R^2$ can hide a model that behaves erratically between data points. Plotting the curve directly against a scatter plot of your data instantly reveals if the curve is physically realistic or if it is suffering from wild oscillations.
