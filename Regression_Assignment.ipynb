1. What is Simple Linear Regression?

Simple Linear Regression is a supervised parametric algorithm used to model the linear relationship between a single independent variable ($X$) and a single continuous dependent variable ($Y$).

2. What are the key assumptions of Simple Linear Regression?

a. Linearity: The relationship between $X$ and the mean of $Y$ is linear.

b. Independence: Observations and residuals are independent of each other (no autocorrelation).

c. Homoscedasticity: Residuals have constant variance across all levels of $X$.

d. Normality: Residuals are normally distributed with a mean of zero.

3. What does the coefficient $m$ represent in the equation $Y = mX + c$?

The coefficient $m$ represents the slope of the regression line. It quantifies the expected change in the dependent variable ($Y$) for every one-unit increase in the independent variable ($X$).

4. What does the intercept $c$ represent in the equation $Y = mX + c$?The intercept $c$ is the value where the regression line crosses the vertical axis. It represents the expected value of $Y$ when $X = 0$.

5. How do we calculate the slope $m$ in Simple Linear Regression?

Using the Ordinary Least Squares (OLS) formula:$$m = \frac{\sum_{i=1}^{n} (X_i - \bar{X})(Y_i - \bar{Y})}{\sum_{i=1}^{n} (X_i - \bar{X})^2} = \frac{\text{Cov}(X, Y)}{\text{Var}(X)}$$Where $\bar{X}$ and $\bar{Y}$ are the sample means of $X$ and $Y$.

6. What is the purpose of the least squares method in Simple Linear Regression?

The Ordinary Least Squares (OLS) method finds the optimal line of best fit by minimizing the Sum of Squared Residuals (SSR):$$\text{SSR} = \sum_{i=1}^{n} (Y_i - \hat{Y}_i)^2$$This ensures the selected line minimizes the overall prediction error across the dataset.

7. How is the coefficient of determination ($R^2$) interpreted in Simple Linear Regression?

$R^2$ indicates the proportion of total variance in the dependent variable ($Y$) explained by the independent variable ($X$). An $R^2$ of $0.85$ means $85\%$ of the variation in $Y$ is accounted for by $X$, with values bounded between $0$ and $1$.

8. What is Multiple Linear Regression?

Multiple Linear Regression extends simple linear regression by modeling the relationship between two or more independent variables ($X_1, X_2, \dots, X_p$) and a single continuous dependent variable ($Y$):$$Y = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + \dots + \beta_p X_p + \epsilon$$

9. What is the main difference between Simple and Multiple Linear Regression?

a. Simple Linear Regression: Uses exactly one predictor variable ($X$); fits a 2D line.

b. Multiple Linear Regression: Uses two or more predictor variables ($X_1, \dots, X_p$); fits a multi-dimensional plane or hyperplane.

10. What are the key assumptions of Multiple Linear Regression?

Linearity: The relationship between predictors and the target variable is linear in parameters.

No Multicollinearity: Independent variables are not highly correlated with each other.

Independence of Errors: Residuals are independent (no autocorrelation).

Homoscedasticity: Constant error variance across all predictions.

Normality of Residuals: Residuals follow a normal distribution, $\epsilon \sim \mathcal{N}(0, \sigma^2)$.

11. What is heteroscedasticity, and how does it affect the results of a Multiple Linear Regression model?

Heteroscedasticity occurs when the variance of the residuals is non-constant across fitted values.

Effect: Coefficients remain unbiased, but the standard errors are biased and incorrect. This invalidates hypothesis tests ($t$-tests, $F$-tests), p-values, and confidence intervals.

12. How can you improve a Multiple Linear Regression model with high multicollinearity?

Remove redundant features with a high Variance Inflation Factor ($\text{VIF} > 5$ or $10$).Combine correlated features using domain knowledge or dimensionality reduction (e.g., PCA).Apply regularized regression techniques (Ridge, Lasso, or Elastic Net) that penalize coefficient magnitudes.

13. What are some common techniques for transforming categorical variables for use in regression models?

One-Hot Encoding / Dummy Encoding: Converts nominal categories into binary $(0/1)$ vectors, dropping the first column to prevent the dummy variable trap.

Ordinal Encoding: Maps ordered categories to ranked integers (e.g., Low $= 1$, Medium $= 2$, High $= 3$).

Target / Frequency Encoding: Replaces categories with the mean target value or observation frequency (used primarily with high-cardinality data).

14. What is the role of interaction terms in Multiple Linear Regression?

Interaction terms model scenarios where the effect of one predictor variable ($X_1$) on the target ($Y$) depends on the magnitude of another predictor ($X_2$). They are introduced as product terms:$$Y = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + \beta_3 (X_1 \times X_2) + \epsilon$$

15. How can the interpretation of intercept differ between Simple and Multiple Linear Regression?

Simple Regression: The expected value of $Y$ when the single variable $X = 0$.

Multiple Regression: The expected value of $Y$ when all included predictor variables simultaneously equal zero ($X_1 = 0, X_2 = 0, \dots, X_p = 0$).

16. What is the significance of the slope in regression analysis, and how does it affect predictions?

The slope defines the magnitude and direction of the linear relationship. A steeper positive slope means predictions increase rapidly with increases in $X$; a negative slope decreases predictions; and a slope near zero indicates that the predictor provides negligible value for predicting $Y$.

17. How does the intercept in a regression model provide context for the relationship between variables?

The intercept sets the baseline starting point of the model. If zero is within the realistic range of inputs (e.g., base salary with zero years of experience), it carries practical meaning. If zero is impossible or out of bounds (e.g., zero height/weight), it acts purely as a mathematical anchor to position the line correctly.

18. What are the limitations of using $R^2$ as a sole measure of model performance?

$R^2$ mechanically increases or stays constant whenever new predictors are added, even if they are pure noise.It does not signal overfitting.A high $R^2$ can still occur in models that violate linear regression assumptions or miss underlying non-linear patterns.

19. How would you interpret a large standard error for a regression coefficient?

A large standard error indicates high uncertainty and low precision in the coefficient's estimate. This typically yields a lower $t$-statistic, a wider confidence interval, and a higher p-value, making it difficult to conclude that the predictor has a statistically significant effect.

20. How can heteroscedasticity be identified in residual plots, and why is it important to address it?

Identification: Observed as a non-random, funnel-, cone-, or fan-shaped spread in a plot of Residuals vs. Fitted Values.

Why address it: Failing to address it produces unreliable p-values and confidence intervals. It can be corrected using log/power transformations on $Y$ or employing Robust Standard Errors (HC errors).

21. What does it mean if a Multiple Linear Regression model has a high $R^2$ but low adjusted $R^2$?

It indicates that the model contains too many useless or redundant predictor variables that do not contribute meaningful explanatory power. The adjusted $R^2$ penalizes the addition of non-informative variables, exposing the model's overfitting.

22. Why is it important to scale variables in Multiple Linear Regression?

Interpretability: Allows direct comparison of feature importance via standardized coefficients ($\beta$).

Optimization: Accelerates convergence for iterative algorithms like Gradient Descent.

Regularization: Required for Ridge, Lasso, and Elastic Net, where penalty terms ($\lambda \sum \beta^2$ or $\lambda \sum \vert{}\beta\vert{}$) are sensitive to the absolute scale of features.

23. What is polynomial regression?

Polynomial regression is a form of regression analysis in which the relationship between the independent variable $X$ and the dependent variable $Y$ is modeled as an $n$-th degree polynomial. It remains linear in terms of its parameters ($\beta$).

24. How does polynomial regression differ from linear regression?

Linear Regression: Models linear, straight-line relationships ($Y = \beta_0 + \beta_1 X$).

Polynomial Regression: Models curvilinear patterns by introducing higher-degree powers ($X^2, X^3, \dots, X^n$), producing curved fit lines while still being solvable via standard linear methods.

25. When is polynomial regression used?

It is used when exploratory data analysis or residual plots reveal non-linear patterns (such as curves, peaks, or diminishing returns) that a straight line cannot adequately capture, while still maintaining the structure of linear estimators.

26. What is the general equation for polynomial regression?

For a single predictor $X$ raised to degree $n$:$$Y = \beta_0 + \beta_1 X + \beta_2 X^2 + \beta_3 X^3 + \dots + \beta_n X^n + \epsilon$$

27. Can polynomial regression be applied to multiple variables?

Yes. In multiple polynomial regression, features include powers of each variable and interaction terms between them (e.g., $X_1^2, X_2^2, X_1 X_2$).

28. What are the limitations of polynomial regression?Highly prone to overfitting when degree $n$ is chosen too high.Highly sensitive to outliers, especially near the boundaries of the data.Poor extrapolation capabilities; predictions outside the observed domain often diverge drastically.

29. What methods can be used to evaluate model fit when selecting the degree of a polynomial?

$k$-Fold Cross-Validation: Tracking training vs. validation error (MSE/RMSE) to find the degree that minimizes test error.

Adjusted $R^2$: Evaluating fit while penalizing extra polynomial degrees.

Information Criteria: Using Akaike Information Criterion (AIC) or Bayesian Information Criterion (BIC) to balance likelihood against parameter count.

30. Why is visualization important in polynomial regression?

Visualization (plotting data points against fitted curves) lets you directly check for the Runge phenomenon (extreme boundary oscillations), assess whether the curve tracks noise rather than signal, and visually confirm whether the model overfits or underfits.31. How is polynomial regression implemented in Python?It is typically implemented using Scikit-Learn's PolynomialFeatures transformer paired with LinearRegression inside a Pipeline

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures

# Sample data
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([1.2, 3.9, 9.1, 16.2, 25.1])

# Degree-2 polynomial regression pipeline
model = make_pipeline(PolynomialFeatures(degree=2), LinearRegression())
model.fit(X, y)

# Predictions
y_pred = model.predict(X)
```
